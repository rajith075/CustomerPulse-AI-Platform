# Phase 4.2 - Enterprise Feature Engineering

## Objective

Create intelligent business features from raw customer data.

These features will improve machine learning performance and provide valuable business insights.

Business Goal:

Transform customer behaviour into measurable AI metrics.

```python
import numpy as np

# Normalize values between 0 and 1

usage = df_clean["Usage Frequency"] / df_clean["Usage Frequency"].max()

payment = 1 - (df_clean["Payment Delay"] / df_clean["Payment Delay"].max())

support = 1 - (df_clean["Support Calls"] / df_clean["Support Calls"].max())

tenure = df_clean["Tenure"] / df_clean["Tenure"].max()

df_clean["Health Score"] = (
    usage * 35 +
    payment * 25 +
    support * 20 +
    tenure * 20
).round(2)
```

```python
df_clean[["Health Score"]].head()
```

```python
engagement = (
    (df_clean["Usage Frequency"] / df_clean["Usage Frequency"].max()) * 70
    +
    (1 - df_clean["Last Interaction"] / df_clean["Last Interaction"].max()) * 30
)

df_clean["Engagement Score"] = engagement.round(2)
```

```python
df_clean[["Engagement Score"]].head()
```

```python
payment_score = (
    1 - df_clean["Payment Delay"] / df_clean["Payment Delay"].max()
)

df_clean["Payment Reliability Score"] = (
    payment_score * 100
).round(2)
```

```python
df_clean[["Payment Reliability Score"]].head()
```

```python
support_score = (
    df_clean["Support Calls"] /
    df_clean["Support Calls"].max()
)

df_clean["Support Severity Score"] = (
    support_score * 100
).round(2)

print("✅ Support Severity Score Created Successfully")
print(df_clean[["Support Severity Score"]].head())
```

```python
df_clean["Estimated CLV"] = (
    df_clean["Total Spend"] *
    (df_clean["Tenure"] / 12)
).round(2)

print("✅ Estimated CLV Created Successfully")
print(df_clean[["Estimated CLV"]].head())
```

```python
df_clean["Revenue At Risk"] = np.where(
    df_clean["Churn"] == 1,
    df_clean["Total Spend"] * 12,
    0
).round(2)

print("✅ Revenue At Risk Created Successfully")
print(df_clean[["Revenue At Risk"]].head())
```

```python
def risk(score):

    if score >= 80:
        return "Low"

    elif score >= 60:
        return "Medium"

    elif score >= 40:
        return "High"

    else:
        return "Critical"

df_clean["Risk Category"] = df_clean["Health Score"].apply(risk)

print("✅ Risk Category Created Successfully")
print(df_clean[["Health Score", "Risk Category"]].head())
```

```python
def segment(row):

    if row["Estimated CLV"] > 3000:
        return "VIP"

    elif row["Health Score"] >= 75:
        return "Loyal"

    elif row["Health Score"] >= 50:
        return "Regular"

    else:
        return "At Risk"

df_clean["Customer Segment"] = df_clean.apply(segment, axis=1)

print("✅ Customer Segment Created Successfully")
print(df_clean[["Estimated CLV", "Health Score", "Customer Segment"]].head())
```

```python
def priority(row):

    if row["Risk Category"] == "Critical":
        return "Immediate"

    elif row["Risk Category"] == "High":
        return "High"

    elif row["Risk Category"] == "Medium":
        return "Normal"

    else:
        return "Low"

df_clean["Customer Priority"] = df_clean.apply(priority, axis=1)

print("✅ Customer Priority Created Successfully")
print(df_clean[["Risk Category", "Customer Priority"]].head())
```

```python
print("=" * 80)
print("🎉 FEATURE ENGINEERING COMPLETED SUCCESSFULLY")
print("=" * 80)

print("\nDataset Shape:")
print(df_clean.shape)

print("\nNew Columns Added:")

new_columns = [
    "Health Score",
    "Engagement Score",
    "Payment Reliability Score",
    "Support Severity Score",
    "Estimated CLV",
    "Revenue At Risk",
    "Risk Category",
    "Customer Segment",
    "Customer Priority"
]

for col in new_columns:
    print(f"✅ {col}")

print("\nPreview of Engineered Dataset:")
print(df_clean.head())
```

# Phase 5.2 - Enterprise ML Model Factory

## Objective

Automatically train, evaluate, compare, and rank multiple machine learning models.

Business Goal

Identify the most accurate model for predicting customer churn while ensuring scalability and explainability.

```python
import time

import pandas as pd

from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    ExtraTreesClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from xgboost import XGBClassifier

from lightgbm import LGBMClassifier

from catboost import CatBoostClassifier
```

```python
models = {

    "Logistic Regression": LogisticRegression(max_iter=1000),

    "Decision Tree": DecisionTreeClassifier(random_state=42),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=200,
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        eval_metric="logloss",
        random_state=42
    ),

    "LightGBM": LGBMClassifier(
        random_state=42
    ),

    "CatBoost": CatBoostClassifier(
        verbose=False,
        random_state=42
    )
}
```

```python
results = []
```

```python
for name, model in models.items():

    print("=" * 80)
    print(f"🚀 Training {name}")
    print("=" * 80)

    start = time.time()

    model.fit(X_train, y_train)

    end = time.time()

    y_pred = model.predict(X_test)

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:
        y_prob = y_pred

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(y_test, y_pred)

    recall = recall_score(y_test, y_pred)

    f1 = f1_score(y_test, y_pred)

    roc = roc_auc_score(y_test, y_prob)

    training_time = round(end - start, 2)

    results.append({

        "Model": name,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1 Score": f1,

        "ROC AUC": roc,

        "Training Time (s)": training_time

    })

    print(f"✅ Finished {name}")
```

```python
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="ROC AUC",
    ascending=False
)

results_df
```

```python
results_df.style.format({

    "Accuracy": "{:.4f}",

    "Precision": "{:.4f}",

    "Recall": "{:.4f}",

    "F1 Score": "{:.4f}",

    "ROC AUC": "{:.4f}",

    "Training Time (s)": "{:.2f}"

}).background_gradient(cmap="Greens")
```

```python
best_model = results_df.iloc[0]

print("=" * 80)
print("🏆 BEST MODEL")
print("=" * 80)

print(best_model)
```

```python
results_df.to_csv(
    "../../datasets/final/model_comparison.csv",
    index=False
)

print("✅ Model comparison saved successfully.")
```
