```python
print("=" * 70)
print("PRODUCTION FEATURE SELECTION")
print("=" * 70)

features = [

    "Age",

    "Gender",

    "Tenure",

    "Usage Frequency",

    "Support Calls",

    "Payment Delay",

    "Subscription Type",

    "Contract Length",

    "Total Spend",

    "Last Interaction"

]

target = "Churn"

X = df[features]

y = df[target]

print()

print("Selected Features")

for feature in features:

    print("✅", feature)

print()

print("Target Column")

print(target)

print()

print("Feature Matrix Shape :", X.shape)

print("Target Shape :", y.shape)
```

```python
print("=" * 70)
print("ENCODING CATEGORICAL FEATURES")
print("=" * 70)

X = pd.get_dummies(

    X,

    drop_first=True,

    dtype=int

)

print()

print("Encoded Dataset Shape")

print(X.shape)

print()

print("Encoded Columns")

print(X.columns.tolist())
```

```python
print("=" * 70)
print("TRAIN TEST SPLIT")
print("=" * 70)

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)

print()

print("Training Features :", X_train.shape)

print("Testing Features :", X_test.shape)

print()

print("Training Labels :", y_train.shape)

print("Testing Labels :", y_test.shape)
```

```python
print("=" * 80)
print("🚀 CUSTOMERPULSE AI - MODEL FACTORY")
print("=" * 80)

models = {

    "Logistic Regression": LogisticRegression(max_iter=1000),

    "Decision Tree": DecisionTreeClassifier(random_state=42),

    "Random Forest": RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    ),

    "Extra Trees": ExtraTreesClassifier(
        random_state=42,
        n_jobs=-1
    ),

    "XGBoost": XGBClassifier(
        random_state=42,
        eval_metric="logloss"
    ),

    "LightGBM": LGBMClassifier(
        random_state=42,
        verbose=-1
    ),

    "CatBoost": CatBoostClassifier(
        random_state=42,
        verbose=False
    )

}

results = []

trained_models = {}

print()

print(f"Total Models : {len(models)}")

print()
```

```python
import time
print("=" * 80)
print("🏋️ TRAINING MODELS")
print("=" * 80)

for model_name, model in models.items():

    print(f"\nTraining {model_name}...")

    start = time.time()

    model.fit(X_train, y_train)

    training_time = time.time() - start

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(y_test, y_pred)

    recall = recall_score(y_test, y_pred)

    f1 = f1_score(y_test, y_pred)

    roc = roc_auc_score(y_test, y_pred)

    trained_models[model_name] = model

    results.append({

        "Model": model_name,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1 Score": f1,

        "ROC AUC": roc,

        "Training Time (s)": round(training_time,2)

    })

    print(f"✅ {model_name} Completed")
```

```python
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(

    by="Accuracy",

    ascending=False

).reset_index(drop=True)

print("=" * 80)
print("🏆 MODEL LEADERBOARD")
print("=" * 80)

display(results_df.style.background_gradient(cmap="Greens"))
```

```python
print("=" * 80)
print("🚀 SELECTING PRODUCTION MODEL")
print("=" * 80)

PRODUCTION_MODEL_NAME = "CatBoost"

production_model = trained_models[PRODUCTION_MODEL_NAME]

print()

print("Production Model Selected:")
print(PRODUCTION_MODEL_NAME)
```

```python
from pathlib import Path

MODEL_DIR = PROJECT_ROOT / "backend" / "saved_models"

MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("Model Directory")

print(MODEL_DIR)
```

```python
import joblib

MODEL_PATH = MODEL_DIR / "customerpulse_catboost.pkl"

joblib.dump(
    production_model,
    MODEL_PATH
)

print()

print("✅ Production Model Saved")

print(MODEL_PATH)
```

```python
FEATURE_PATH = MODEL_DIR / "feature_names.pkl"

joblib.dump(
    X.columns.tolist(),
    FEATURE_PATH
)

print()

print("✅ Feature Names Saved")

print(FEATURE_PATH)
```

```python
metadata = {

    "project": "CustomerPulse AI",

    "model_name": PRODUCTION_MODEL_NAME,

    "algorithm": "CatBoostClassifier",

    "version": "1.0.0",

    "created_on": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),

    "training_samples": len(X_train),

    "testing_samples": len(X_test),

    "feature_count": len(X.columns),

    "feature_names": X.columns.tolist()

}

with open(

    MODEL_DIR / "model_metadata.json",

    "w"

) as f:

    json.dump(

        metadata,

        f,

        indent=4

    )

print()

print("✅ Metadata Saved")
```

```python
best_row = results_df[
    results_df["Model"] == PRODUCTION_MODEL_NAME
].iloc[0]

metrics = {

    "accuracy": float(best_row["Accuracy"]),

    "precision": float(best_row["Precision"]),

    "recall": float(best_row["Recall"]),

    "f1_score": float(best_row["F1 Score"]),

    "roc_auc": float(best_row["ROC AUC"]),

    "training_time": float(best_row["Training Time (s)"])

}

with open(

    MODEL_DIR / "model_metrics.json",

    "w"

) as f:

    json.dump(

        metrics,

        f,

        indent=4

    )

print()

print("✅ Metrics Saved")
```
